```python
import pandas as pd
```

```python
file_path = "Messy_Employee_dataset.csv"

df = pd.read_csv(file_path, encoding="utf-8", sep=",", dtype={
    "employee_id": "string",
    "age": "float",
    "salary":"float"
}
)

df.head()
```

```python
df.info()
df.isnull().sum()
df.duplicated().sum()
```

```python
for col in df.columns:
    print(f"\n{col}")
    print(df[col].unique()[:10])
```

```python
df.columns = df.columns.str.strip().str.lower()
```

```python
df["join_date"] = pd.to_datetime(df["join_date"], errors="coerce")
```

```python
df["salary"] = (
    df["salary"]
    .astype(str)
    .str.replace(",","")
    .replace("N/A", None)
)

df["salary"] = pd.to_numeric(df["salary"], errors="coerce")
```

```python
df[["depertment", "region"]] = df["department_region"].str.split("-", expand=True)
```

```python
df["status"] = df["status"].str.strip().str.lower()

df["performance_score"] = df["performance_score"].str.strip().str.lower()
```

```python
df["age"] = df["age"].fillna(df["age"].median())

df["salary"] = df["salary"].fillna(df["age"].median())
```

```python
df = df.drop_duplicates()
```

```python
df.info()
df.isnull().sum()
df.describe()
```

```python
df.to_csv("Cleaned_employee_dataset.csv", index=False)
```
